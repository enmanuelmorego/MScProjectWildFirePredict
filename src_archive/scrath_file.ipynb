## Sampling Procedure
The function `run_tabular` prepares all the tabular data sources into a single data frame where each day covered in the date range is assigned a `grid_id` and the composite key is created. This dataset also contains the `fire_lbl` value to indicate whether specific locations at specific times experienced wildfires or not. 

The included columns are: 
`['grid_id', 'x_coord', 'y_coord', 'geometry', 'date', 'viirs_n', 'frp_max', 'frp_mean', 'fire_lbl', 'fwi_max', 'fwi_mean', 'composite_key']`

### Steps
- Create a data frame `df_fire` containing only rows where `fire_lbl` = `True`
- Create an empty dictionary `samples_dict` as: 

```
{fire_lbl_comp_key: [],
 temporal_sample_comp_key: [],
 spatial_sample_comp_key: []}
 ```

 - Create two functions, `extract_temporal_sample()` and `extract_spatial_sample()` which are used to extract given samples
 - Pass `df_fire`, `samples_dict` thru both functions
 - Each function iterates thru `df_fire` and adds `composite_key` of `df_fire` row to `samples_dict['fire_lbl_comp_key']` 
 - Then for each corresponding sample, the matching value gets added to `temporal_sample_comp_key` and `spatial_sample_comp_key` (if no value is found, add flag 'sample_not_found')

 This allows to generate a sampling report to measure whether samples where found or not for a given `composite_id`

 Then all of these `composite_ids` are transformed into a single vector, or column, and used to filter the data generated by `run_tabular` and this is then stored as the sampled data set

 Please note that:
 Composite ids cannot be repeated. The sample prodcedure ensures that this is the case 

````python
df_fire = df.copy()
df_fire = df_fire[df_fire['fire_lbl'] == True]

# Extract dataset that has never seen a fire 
df_nofire = df[~df['grid_id'].isin(df_fire['grid_id'])]

df_fire.columns
````

````python
import pandas as pd

from typing import cast
import importlib
import src.sampling.sampling_functions as sf
importlib.reload(sf)

samples_dict = {'fire_lbl_comp_key'       : [],
                'temporal_sample_comp_key': [],
                'spatial_sample_comp_key' : [],
                'spatial_sample_dist'     : [],
                'used_comp_keys'          : set()}

for row in df_fire.itertuples():

    composite_tgt = row.composite_key
    grid_tgt      = cast(int, row.grid_id)
    date_tgt      = cast(pd.Timestamp, row.date)
    x_tgt         = cast(float, row.x_coord)
    y_tgt         = cast(float, row.y_coord)
    used_comp_key = samples_dict['used_comp_keys']

    df_to_sample_temporal = df[df['grid_id'] == grid_tgt]
    df_to_sample_spatial  = df_nofire[df_nofire['date'] == date_tgt]

    temp_sample                 = sf.extract_temporal_sample(df_to_sample_temporal, grid_tgt, date_tgt, used_comp_key)
    spatial_sample, sample_dist = sf.extract_spatial_sample(df_to_sample_spatial, x_tgt, y_tgt)
    
    
    samples_dict['fire_lbl_comp_key'].append(composite_tgt)
    samples_dict['temporal_sample_comp_key'].append(temp_sample)
    samples_dict['spatial_sample_comp_key'].append(spatial_sample)
    samples_dict['spatial_sample_dist'].append(sample_dist)
    samples_dict['used_comp_keys'].add(temp_sample)
    samples_dict['used_comp_keys'].add(spatial_sample)

    
````
